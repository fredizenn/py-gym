````python
import threading
import time


files = ["file1", "file2", "file3", "file4", "file5"]

def download(file_name):
    print(f"\nDownloading {file_name}")
    time.sleep(2)
    print(f"\nDownload {file_name} finished")
    
threads = []

for i in files:
    t = threading.Thread(target=download, args=(i,))
    threads.append(t)
    t.start()

for t in threads:
    t.join()
````

## **Problem 2: Multi-processing Computation**

Create a program that calculates the square of numbers from 1 to 8.

**Requirements**:

- Simulate a one-second delay for each computation

- Run computations in parallel

**Questions**:
- How many processes are created?

- Does the runtime change when compared with a sequential program?


```python
# Starter Code

import time

    pass

```
----

````python
import multiprocessing


numbers = list(range(1, 9))
def compute_square(number):
    time.sleep(1)
    return number ** 2


def multiprocess_square_number():
    start = time.perf_counter()
    p1 = multiprocessing.Process(target=compute_square, args=numbers,)
    p2 = multiprocessing.Process(target=compute_square, args=numbers,)
    p1.start(); p2.start()
    p1.join(); p2.join()
    
    print(f"[Processing | CPU] {time.perf_counter() - start:.2f}s")

print("\nRunning processing tasks:")
multiprocess_square_number()  
````

## **Problem 3: Asynchronous Program**
Create an asynchronous program with three tasks:
- `load_data()`
- `process_data()`
- `save_data()`

**Requirements**:

- Each task should pause for two seconds using `asyncio.sleep`

- Run all tasks concurrently

- Print start and finish messages for each task
```python
# Starter code

    pass

    pass

async def save_data():
    pass

async def main():
    pass

```
---

````python
import asyncio


async def load_data():
    print("Loading data")
    await asyncio.sleep(2)
    print("Loading complete")
    
async def process_data():
    print("Processing data")
    await asyncio.sleep(2)
    print("Processing complete")
    
async def save_data():
    print("saving")
    await asyncio.sleep(2)
    print("saved")
    
    
async def main():
    task1 = asyncio.create_task(load_data())
    task2 = asyncio.create_task(process_data())
    task3 = asyncio.create_task(save_data())
    
    await asyncio.gather(task1, task2, task3)
    
await main()
````

## **Problem 5: Async HTTP Requests**

Create a program that downloads the content of three websites concurrently.

**Requirements**:

- use `aiohttp`

- create asynchronous requests

- print the size of each downloaded page

**URLs**:

- https://example.com
- https://python.org
- https://github.com

```python
# Starter code
import asyncio
import aiohttp

urls = [
    "https://example.com",
    "https://python.org",
    "https://github.com"
]

async def fetch(session, url):
    pass

async def main():
    pass

asyncio.run(main())
```
---

````python
import aiohttp


urls = [
    "https://example.com",
    "https://python.org",
    "https://github.com"
]


async def fetch(session, url):
    async with session.get(url) as response:
       content = await response.text()
       return url, len(content)
        
async def main():
    async with aiohttp.ClientSession() as session:
        tasks = [asyncio.create_task(fetch(session, url)) for url in urls]
        results = await asyncio.gather(*tasks)

    for url, size in results:
        print(f"{url}: {size} characters")

asyncio.run(main())
````
